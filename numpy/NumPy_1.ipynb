9. Stacking

```python

a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)
```

```python
a4
```

```python
a5
```

```python
#horizontal stacking
np.hstack((a4,a5))
```

```python
#vertical stacking
np.vstack((a4,a5))
```
