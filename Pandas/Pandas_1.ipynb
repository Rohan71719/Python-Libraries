```python
import pandas as pd
import numpy as np
```

1.SERIES FROM LISTS

```python
# string
country=['India','pak','nepal','USA']
pd.Series(country)
```

```python
# integers
runs=['13','20','30','50']
runs_series=pd.Series(runs)
runs_series
```

```python
#custom index
marks=[67,57,89,100]
subjects=['maths','english','science','bengali']

pd.Series(marks, index=subjects)
```

```python
# setting a name
marks=pd.Series(marks, index=subjects, name="Marks of Rohan")
marks
```

2. SERIES FROM DICTIONARY

```python
marks={
    'maths':67,
    'english': 57,
    'math': 90,
    'bengali': 100
}

marks_series=pd.Series(marks, name="rohan")
marks_series
```
